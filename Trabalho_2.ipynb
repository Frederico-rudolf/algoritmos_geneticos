Busca de uma rota curta por 20 cidades, com retorno à origem. As funções estão comentadas por etapas em `main.py`.

## 1. Preparar os dados

Execute este notebook na pasta do projeto, usando apenas a biblioteca padrão do Python. O arquivo enviado é texto do Octave.

```python
import random
from main import executar, desenhar_caminho, carregar_cidades, desenhar_custos

x, y = carregar_cidades("cidades.mat")
print("Número de cidades:", len(x))
```

## 2. Executar o algoritmo

População de 20, 10.000 gerações, retenção das 10 melhores, roleta por ranking, crossover cycle e mutação em todos os indivíduos. A semente permite reproduzir esta execução.

```python
random.seed(42)
historico = []
populacao_inicial, populacao_final, melhor_solucao, melhor_custo = executar(
    x, y, numero_geracoes=10000, tamanho_populacao=20, historico=historico
)
```

## 3. Apresentar os resultados

O melhor resultado histórico é guardado separadamente: a mutação também altera os sobreviventes.

```python
print("Tamanho da população:", len(populacao_inicial))
print("Número de cidades:", len(x))

for titulo, populacao in [("População inicial", populacao_inicial),
                          ("População final", populacao_final)]:
    print("\n" + titulo)
    for i, rota in enumerate(populacao, start=1):
        print(f"Indivíduo {i}: {rota}")

print(f"\nMelhor custo: {melhor_custo:.6f}")
print("Melhor solução:", melhor_solucao)
```

## 4. Mostrar o percurso

A primeira cidade é repetida no desenho para fechar o caminho. Abra `resultados/melhor_caminho.svg` no navegador para visualizar. O algoritmo não garante o ótimo global.

```python
desenhar_caminho(x, y, melhor_solucao, melhor_custo,
                 "resultados/melhor_caminho.svg", mostrar=False)
print("Gráfico salvo em resultados/melhor_caminho.svg")
```

## 5. Evolução do melhor custo
O gráfico inclui a população inicial (geração 0) e cada uma das 10.000 gerações. Ele mostra o melhor custo histórico, que diminui ou permanece igual. Os dois gráficos são SVGs gerados sem Matplotlib ou uma função `plot`.


```python
desenhar_custos(historico, "resultados/melhor_custo.svg", mostrar=False)
print("Gráfico salvo em resultados/melhor_custo.svg")
```
